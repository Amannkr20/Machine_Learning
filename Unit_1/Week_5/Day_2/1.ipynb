```python
class Largest:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def find_largest(self):
        if self.a>=self.b and self.a>=self.c:
            return self.a
        elif self.b>=self.b and self.b>=self.c:
            return self.b
        else:
            return self.c
obj=Largest(12,45,67)
print("The largest number is",obj.find_largest())

    
```
