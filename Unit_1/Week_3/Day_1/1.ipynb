```python
import pandas as pd
```

```python
data={
    "name":['Bhavya','Aman','Pawan'],
    "emp_id":[100,200,300],
    "unit":[10,20,30],
    "City":['gurugram','raipur','banaras'],
    "Cp":[100,200,300],
    "Sp":[200,100,400],
    "Diff": [100,100,100],
    "Total_cost": [100,100,300]


}

df=pd.DataFrame(data)
df
```
