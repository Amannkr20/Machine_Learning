```python
import numpy as np
```

```python
X=np.array([[40,5],[30,6],[25,5.6],[120,25],[110,35],[120,30],[140,25]])
Y=np.array([1,1,1,0,0,0,0])
```

```python
from sklearn.neighbors import KNeighborsClassifier

import matplotlib.pyplot as plt

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, Y)

x_min, x_max = X[:, 0].min() - 10, X[:, 0].max() + 10
y_min, y_max = X[:, 1].min() - 5, X[:, 1].max() + 5
xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 1),
    np.arange(y_min, y_max, 1)
)

regions = knn.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, regions, alpha=0.25, cmap="coolwarm")
plt.scatter(X[:, 0], X[:, 1], c=Y, cmap="coolwarm", edgecolor="black")
plt.xlabel("PetalLengthCm")
plt.ylabel("PetalWidthCm")
plt.title("KNN classification (k=3)")
plt.show()
```
